```r
# importing the tidymodels and repr packages
library(tidymodels)
library(repr)

# setting the maximum rows displayed for a tibble as 6
options(repr.matrix.max.rows = 6)

# dropping all observations with missing values for the conf_wins_pct, reg_wins, or bid columns and assigning the new
# data frame to ncaa_data_no_missing_values. This is because my analysis won't run if there are missing values
ncaa_data_no_missing_values <- ncaa_data_tidy |>
    drop_na()

# displaying ncaa_data_no_missing_values
ncaa_data_no_missing_values

# splitting the data set into a training set and a testing set. 75% of the observations go into the training set and 
# the other 25% go into the testing set. I also do strata = bid in order to have approximately the same proportions of 
# the "at-large" and "auto" bid types in both training and testing sets

ncaa_split <- initial_split(ncaa_data_no_missing_values, prop = 0.75, strata = bid)
ncaa_train <- training(ncaa_split)
ncaa_test <- testing(ncaa_split)

# displaying the training and testing sets
ncaa_train
ncaa_test


```

```r
# creating the model specification for tuning
knn_tune <- nearest_neighbor(weight_func = "rectangular", neighbors = tune()) |>
    set_engine("kknn") |>
    set_mode("classification")

# displaying the model specification
knn_tune

# creating the recipe for my analysis where I specify that "bid" is the response variable and that "conf_wins_pct" and
# "reg_wins" are my 2 predictor variables. I also scale and center both my predictor variables so that they are on 
# comparable scales. Therefore, one predictor variable that has values at a larger scale won't have more influence 
# on the predictions than the other predictor variable.
ncaa_recipe <- recipe(bid ~ conf_wins_pct + reg_wins, data = ncaa_train) |>
    step_scale(all_predictors()) |>
    step_center(all_predictors())

# displaying the recipe
ncaa_recipe

# creating the ncaa_vfold object which allows me to do 5-fold cross validation
ncaa_vfold <- vfold_cv(ncaa_train, v = 5, strata = bid)

# displaying the 5-fold cross validation
print(ncaa_vfold)

# the k_values tibble contains the different K-values I will try out in order to determine which K-value will be the best
# one to use for my model. I'm trying out k-values ranging from 1 to 201 (incrementing by 4) to figure out which k-value 
# ends up with the highest accuracy. I'm also only trying out odd k-values in order to avoid ties when predicting 
# the bid type.
k_values <- tibble(neighbors = seq(from = 1, to = 201, by = 4))

# displaying the k-values tibble
k_values

# ncaa_results contains the results after doing 5-fold cross validation on the set of different k-values. I'm mainly
# focusing on the estimated accuracy of those different K-values. 
ncaa_results <- workflow() |>
    add_model(knn_tune) |>
    add_recipe(ncaa_recipe) |>
    tune_grid(resamples = ncaa_vfold, grid = k_values) |>
    collect_metrics() |>
    filter(.metric == "accuracy")

# displaying the 5-fold cross validation results on different values of K
ncaa_results

# creating a scatterplot/lineplot that looks at the relationship between the k-value on the x-axis and the accuracy on
# the y-axis
k_value_vs_accuracy_plot <- ncaa_results |>
    ggplot(aes(x = neighbors, y = mean)) +
    geom_point() + 
    geom_line() + 
    labs(x = "Neighbors", y = "Estimated Accuracy")

# displaying the scatterplot/lineplot
k_value_vs_accuracy_plot


```

```r
# finding the k-value that resulted in the highest estimated accuracy
best_k_value <- ncaa_results |>
    arrange(desc(mean)) |>
    slice(1) |>
    select(neighbors) |>
    pull()

# displaying the best k-value
best_k_value
```

```r
# creating a new model specification but this time with neighbors set to the k-value that had the highest estimated accuracy
# after 5-fold cross validation, which is k = 109
knn_best_spec <- nearest_neighbor(weight_func = "rectangular", neighbors = best_k_value) |>
    set_engine("kknn") |>
    set_mode("classification")

# displaying the new model specification 
knn_best_spec

# training the classification model using k = 109 on the training set
ncaa_fit <- workflow() |>
    add_model(knn_best_spec) |>
    add_recipe(ncaa_recipe) |>
    fit(data = ncaa_train)

# displaying the fit object
ncaa_fit

# predicting on the testing set using the trained classification model and combining the predicted bid types with the 
# conf_wins_pct, reg_wins, and bid columns
ncaa_predictions <- ncaa_fit |>
    predict(ncaa_test) |>
    bind_cols(ncaa_test)

# displaying the predictions
ncaa_predictions




```

```r
# determining the accuracy of the model's predictions on the testing set
ncaa_prediction_accuracy <- ncaa_predictions |>
    metrics(truth = bid, estimate = .pred_class) |>
    filter(.metric == "accuracy")

# displaying the accuracy
ncaa_prediction_accuracy

```
